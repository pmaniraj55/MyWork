```python
#Autograd (compute gradient)

import torch
x=torch.tensor(2.0)
x.requires_grad,x.is_leaf
```

```python
y=3*torch.sigmoid(x)+5
y.requires_grad,y.is_leaf
```

```python
import torch
x = torch.tensor(2.0, requires_grad=True)
x.requires_grad,x.is_leaf

y=3*torch.sigmoid(x)+5

print(y)
y.requires_grad,y.is_leaf

print(x.grad)
print(x.grad_fn)
print(y.grad_fn)

y.backward()
print(x.grad) #dy/dx
```

```python
#Gradient gets accumulated
#x.grad.zero_()
print(x.grad)

y=3*torch.sigmoid(x)+5
y.backward()
x.grad
```

```python
a=torch.rand(2,5,requires_grad=True)
print(a)

b=a*a+a+5
print(b)

c=b.mean()
print(c)

print(a.is_leaf,b.is_leaf,c.is_leaf)

b.retain_grad() #otherwise issue will come in b.grad
print("Before gradient computation a is:", a.grad)
c.backward()
print("After gradient computation a is:", a.grad)

print(b.grad)
```
